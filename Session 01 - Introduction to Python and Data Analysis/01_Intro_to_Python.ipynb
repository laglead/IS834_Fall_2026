## Programming *in* Python vs. analyzing data *in* Python

These are different jobs. A software engineer writes Python to build systems that run unattended for years. You are going to write Python to answer a question about a dataset, look at the answer, and then usually throw the code away.

That difference is why this course teaches a small subset of the language well rather than the whole language badly. You need enough Python to load data, reshape it, and describe it honestly — not enough to build an application.


## Notebooks versus scripts

**Notebooks** (what we use all term) interleave code, output, and prose. You run one cell at a time and see the result immediately, which is exactly what exploratory analysis needs. The cost: because cells can be run in any order, a notebook can end up in a state you cannot reproduce.

**Scripts** (plain `.py` files) run top to bottom, every time, in the same order. That is what production systems use.

Rule of thumb: explore in a notebook, ship in a script.

## Variables and comparison


### Variable assignment

`=` stores something in a variable. The name goes on the left, the value on the right. Read it as "gets", not as "equals" — a single `=` never asks a question.


```python
# Three variables holding three different types of value.
x = 1
y = 2.0
z = 'Brock'
```

```python
# == is the QUESTION "are these equal?". It returns True or False.
# One equals sign stores; two equals signs compare. Mixing them up is the most common beginner bug.
x == y
```

```python
# Python is case sensitive, so these two strings are not the same value.
z2 = 'BROCK'
print(z)
print(z2)
z == z2
```

```python
# type() tells you what kind of value a variable holds -- worth checking constantly.
print(type(x), x)
print(type(y), y)
print(type(z), z)
```

### Inspection

Checking what you actually have, rather than what you assume you have, is most of debugging.


```python
# The quotes matter: this is the TEXT "1.0", not the number 1.0.
# A column of numbers read in as text is the single most common data-loading problem you will hit.
x = "1.0"
print(type(x))
```

```python
# Converting between types is called casting. float() turns the text into a number.
x_number = float(x)
print(type(x_number), x_number)
```

### Comparison operators

Each of these returns a **boolean** — `True` or `False`.


```python
# The six comparisons you will use.
x = 1
y = 2

print(x == y)   # equal to
print(x != y)   # not equal to
print(x < y)    # less than
print(x > y)    # greater than
print(x <= y)   # less than or equal to
print(x >= y)   # greater than or equal to
```

# Lists and Dictionaries

![types](https://k21academy.com/wp-content/uploads/2021/10/data-types.png)

These two are how Python holds *collections* of values. A **list** is an ordered sequence you reach by position. A **dictionary** is a set of labelled values you reach by name.


```python
# A list is written with square brackets. It can hold anything, including a mix of types.
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed_list = ["apple", 42, 3.14, True]

print(fruits)
print(mixed_list)
```

```python
# len() works on lists the same way it works on strings.
print(len(fruits), len(numbers))
```

```python
# Indexing starts at 0, not 1. Negative numbers count back from the end.
fruits = ["apple", "banana", "cherry"]

print(fruits[0])    # first
print(fruits[1])    # second
print(fruits[-1])   # last -- easier than working out len(fruits) - 1
```

```python
# Slicing takes a range. The start is included, the stop is NOT -- [1:3] gives positions 1 and 2.
numbers = [1, 2, 3, 4, 5, 6, 7]

print(numbers[1:3])   # from index 1 up to (not including) 3
print(numbers[:2])    # from the beginning up to index 2
print(numbers[3:])    # from index 3 to the end
```

```python
# Two ways to add a single item: at the end, or at a chosen position.
shopping = ["apple", "banana"]

shopping.append("cherry")          # add ONE item at the end
shopping.append("kiwi")  # add MULTIPLE items at the end
# shopping.insert(0, "avocado")      # add ONE item at a chosen position
shopping.insert(-1, "avocado")

print(shopping)
```

```python
# Basic numeric summaries of a list.
numbers = [3, 1, 4, 1, 5, 9]

print(min(numbers), max(numbers), sum(numbers))
print(sum(numbers) / len(numbers))   # the mean, done by hand

import statistics
print(statistics.median(numbers))   # the mean, done by the statistics package
```

```python
# + on two lists joins them, which is the same result as .extend().
# The difference: + builds a NEW list, .extend() changes the original in place.
print([1, 2] + [3, 4])
```

### Dictionaries

A list answers "what is in position 3?". A dictionary answers "what is the customer's city?" — which is almost always the more useful question.


```python
# Curly braces, and each entry is a key: value pair.
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "interests": ['fine dining', 'concerts', 'hockey']
}
print(person)
print(person["interests"][-1])
print(person["name"][0])
print(person["age"])
```

```python
# Two ways to read a value out.
print(person["name"])           # brackets -- raises an error if the key is missing
print(person.get("city"))       # .get() -- returns None instead of erroring
print(person.get("phone", "not on file"))   # .get() can supply a fallback
print(person.get("address","?"))
```

```python
# The three views of a dictionary you will use most.
print(list(person.keys()))
print(list(person.values()))
print(list(person.items()))     # key/value pairs, which is what you loop over
```

```python
# Assigning to a new key adds it.
person["email"] = "alice@example.com"
print(person["email"])
# print("interests" in person)
```

```python
# del removes a key.
del person["email"]
print("email" in person)
```

```python
# Dictionaries can hold other dictionaries. This is how nested data (JSON, API responses) arrives.
person["grades"] = {"IS834": "A", "IS833": "A-"}
print(person["grades"]["IS834"])
print(person)
```
