# Strings

- `'brock'` `==` `"brock"` are both strings --> Note the use of single or double quotes, either are fine, but we can't mix
- python __is case sensitive!__  `brock` != `Brock` != `BROCK`

__Notes:__
- strings have a length
- strings are sequences so we can `slice` into them (i.e. extract contents by indexing)
- strings have methods to inspect, change case, etc.


```python
# A string is text wrapped in quotes. Single or double both work, as long as you do not mix them in one string.
myname = "Brock"
print(myname)
```

```python
# len() counts the characters in a string.
len(myname)
```

> `len` is common across various python data types!


```python
# .lower() and .upper() hand back a NEW string -- the original variable is untouched.
print(myname.lower())
print(myname.upper())
print(myname)
```

```python
# Store the lowercased version if you want to keep it around.
myname_lower = myname.lower()
myname_lower
```

```python
# Python is case sensitive, so the two spellings are not the same value.
myname == myname_lower
```

```python
# + pastes strings together -- notice that you have to supply the space yourself.
'Brock' + ' ' + 'likes to code in Python'
```

```python
# Two string literals written side by side are joined automatically.
# This only works for literals typed out in your code, never for variables.
'Brock'  ' '  'likes to code in Python'
```

```python
# .count() reports how many times a sequence of characters shows up.
greeting = "Hello IS834"
print(greeting.count("l"))
print(greeting.count("e"))
```

```python
# .startswith() and .endswith() ask a yes-or-no question about the ends of a string.
print(greeting.endswith("834"))
print(greeting.startswith("Hello"))
```

```python
drinking_age = 18
#########################
ages = [18, 22, 75, 3]

for i in ages:
    if i < drinking_age:
        print("Age :" + str(i) + " --> No drinks!")
    else:
        print("Age :" + str(i)+ " --> Serve drinks!")
```

# If Else Logic

![](https://www.learnbyexample.org/wp-content/uploads/python/Python-elif-Statement-Syntax.png)


```python
# A plain if runs its body only when the condition is True, and does nothing otherwise.
number = 10

if number > 5:
    print("The number is greater than 5.")
```

```python
# else gives the alternative path, so exactly one of the two branches always runs.
age = 16

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# elif adds more branches. Python checks them top to bottom and stops at the first True one.
temperature = 75

if temperature > 85:
    print("It's hot outside.")
elif temperature > 65:
    print("The weather is nice.")
else:
    print("It's a bit chilly.")
```

```python
# and requires both conditions to be True; or requires only one of them.
x = 10
y = 20

if not(x > 5) and y > 15:
    print("Both conditions are met.")

if x < 5 or y > 15:
    print("At least one condition is met.")
```

```python
# in checks membership -- it works on lists, strings and dictionary keys.
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("Banana is in the list.")
else:
    print("Banana is not in the list.")
```

# Functions

![](https://www.learnbyexample.org/wp-content/uploads/python/Python-Function-Syntax.png)

- Functions are powerful because they let us repeat the same task over and over.
- We can provide different arguments as inputs for our code to evaluate

> Why cover functions now?  We can use functions to apply (powerful) transformations to the rows across a column in our dataset.


```python
# def names the function; the indented body is what it does. Nothing runs until you call it.
def greet():
    print("Hello, world!")

greet()
```

```python
# A parameter lets the same function handle different inputs.
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person(name="Brock")   # naming the argument makes the call self-documenting
```

```python
def greet(name='Ayra & Aru'):
    print(f'Hi {name}!')

greet()
greet('X')
```

```python
# return hands a value back to you, so it can be stored and reused. print() only shows it on screen.
def add_numbers(a, b):
    return a + b

total = add_numbers(2, 3)
print(total)
```

# Looking Ahead

We are now going to take this fundamentals and move into data analysis with Pandas, which as we will see next week, builds on above!


```python
# pandas and matplotlib both ship with Colab, so they only need importing.
import pandas as pd
import matplotlib.pyplot as plt

# Every data file this notebook uses is read straight from the course repository.
# There is nothing to upload: DATA is the folder those files live in.
DATA = "https://raw.githubusercontent.com/laglead/IS834_Fall_2026/main/data/"
```

```python
# Read the diamonds file straight from the course repository -- nothing to upload.
diamonds = pd.read_csv(DATA + "diamonds.csv")
diamonds.head()
```

```python
# A function that works on one row of the dataset. row['price'] reads a column, exactly like a dictionary key.
def calculate_price_per_carat(row):
    return row['price'] / row['carat']
```

```python
# .apply(axis=1) runs that function once per row and returns the results as a new column.
diamonds['price_per_carat'] = diamonds.apply(calculate_price_per_carat, axis=1)
diamonds[['carat', 'price', 'price_per_carat']].head()
```

```python
# Plot the new column against carat weight to see whether bigger stones cost more per carat.
plt.figure(figsize=(10, 6))
plt.scatter(diamonds['carat'], diamonds['price_per_carat'], alpha=0.3, c='teal')
plt.title('Price per Carat vs. Carat Weight of Diamonds')
plt.xlabel('Carat')
plt.ylabel('Price per Carat')
plt.grid(True, alpha=.1)
plt.show()
```

**Exercise 18.** Write a function `price_band(price)` that returns `"budget"` for prices under 1,000, `"mid"` for prices under 5,000, and `"premium"` otherwise. Test it on a short list of prices with a loop, then use it to build a `band` column on the diamonds data.


```python
# Your code here -- step 1: write price_band() and test it on a few prices
def price_band(price):
    if price < 1000:
        return 'budget'
    elif price < 5000:
        return 'mid'
    else: return 'premium'

prices = [500, 1200, 5200]
for i in prices:
    print(price_band(i))
```

```python
# Your code here -- step 2: apply it to the price column and count the bands
def price_band(row):
    if row['price'] < 1000:
        return 'budget'
    elif row['price'] < 5000:
        return 'mid'
    else: return 'premium'

diamonds['band'] = diamonds.apply(price_band, axis=1)
diamonds[['price','band']].head()
```
